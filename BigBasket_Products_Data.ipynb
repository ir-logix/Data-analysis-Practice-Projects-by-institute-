```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("BigBasket Products.csv")        # Step-1
df.head(12)       # Step-2
```

```python
df.describe(include = 'all')        # Step-3
```

```python
df.info()       # Step-4
```

```python
# Using rating as the proxy for popularity        # Step-5

top_products = (
    df.dropna(subset=['rating'])
      .sort_values('rating', ascending=False)[['product', 'brand', 'sale_price', 'rating']]
      .head(10)
)
print("=== TOP 10 PRODUCTS (by rating) ===")
print(top_products.to_string(index=False))

least_products = (
    df.dropna(subset=['rating'])
      .sort_values('rating', ascending=True)[['product', 'brand', 'sale_price', 'rating']]
      .head(10)
)
print("\n=== LEAST PREFERRED 10 PRODUCTS (by rating) ===")
print(least_products.to_string(index=False))

print("\n=== PRODUCT COUNT BY CATEGORY ===")
print(df['category'].value_counts().head(10))
```

```python
# Discounts on products in percentage       # Step-6

df['discount_pct'] = ((df['market_price'] - df['sale_price']) / df['market_price']) * 100
df['discount_pct'] = df['discount_pct'].round(2)

print("=== TOP 10 MOST DISCOUNTED PRODUCTS ===")
top_discount = (
    df.dropna(subset=['discount_pct'])
      .sort_values('discount_pct', ascending=False)
      [['product', 'market_price', 'sale_price', 'discount_pct']]
      .head(10)
)
print(top_discount.to_string(index=False))

print("\n=== AVERAGE DISCOUNT % BY CATEGORY ===")
print(df.groupby('category')['discount_pct'].mean().sort_values(ascending=False).round(2))
```

```python
df.isnull()       # Step-7  # method-1
```

```python
df.isnull().sum()       # Step-7  # method-2
```

```python
plt.figure(figsize=(10, 4))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing Value Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Check outliers and fill with mean       # Step-8

numeric_cols = ['sale_price', 'market_price', 'rating', 'discount_pct']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(col, "— Outliers:", outlier_count, "|  Bounds:", [(lower.round(2)), (upper.round(2))])

    col_mean = df[col].mean()
    df[col] = np.where((df[col] < lower) | (df[col] > upper), col_mean, df[col])

print("\nOutliers replaced with column mean.")
```

### Create Plots Or Visualizations
__Step-9

```python
# 1. Outliers Modification

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, col in enumerate(numeric_cols):
    df[col].plot(kind='box', ax=axes[i], title=col)
plt.suptitle('Box Plots After Outlier Modification')
plt.tight_layout()
plt.show()
```

```python
# 2. Top 10 categories by product count

plt.figure(figsize=(10, 5))
df['category'].value_counts().head(10).plot(kind='bar', color='steelblue', edgecolor='white')
plt.title('Top 10 Categories by Product Count')
plt.xlabel('Category'); plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout(); plt.show()
```

```python
# 3. Rating distribution

plt.figure(figsize=(8, 4))
sns.histplot(df['rating'].dropna(), bins=20, kde=True, color='coral')
plt.title('Rating Distribution')
plt.xlabel('Rating'); plt.ylabel('Frequency')
plt.tight_layout(); plt.show()
```

```python
# 4. Sale price vs Market price scatter

plt.figure(figsize=(8, 5))
plt.scatter(df['market_price'], df['sale_price'], alpha=0.3, color='teal', s=10)
plt.plot([0, df['market_price'].max()], [0, df['market_price'].max()], 'r--', label='No discount line')
plt.title('Sale Price vs Market Price')
plt.xlabel('Market Price (₹)'); plt.ylabel('Sale Price (₹)')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# 5. Average discount by top 10 categories

top_cats = df['category'].value_counts().head(10).index
avg_disc = df[df['category'].isin(top_cats)].groupby('category')['discount_pct'].mean().value_counts(ascending=False)
plt.figure(figsize=(10, 5))
avg_disc.plot(kind='barh', color='mediumpurple')
plt.title('Average Discount % by Top 10 Categories')
plt.xlabel('Avg Discount %')
plt.tight_layout(); plt.show()
```

```python
# 6. Correlation heatmap

plt.figure(figsize=(6, 4))
corr = df[['sale_price', 'market_price', 'rating', 'discount_pct']].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap')
plt.tight_layout(); plt.show()
```

```python
# 7. Top 10 brands by product count

plt.figure(figsize=(10, 5))
df['brand'].value_counts().head(10).plot(kind='bar', color='darkorange', edgecolor='white')
plt.title('Top 10 Brands by Product Count')
plt.xlabel('Brand'); plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout(); plt.show()
```
